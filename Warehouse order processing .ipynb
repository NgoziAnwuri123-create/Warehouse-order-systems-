```python
from PIL import Image, ImageDraw, ImageFont

# =========================
# WAREHOUSE ORDER STORAGE
# =========================

warehouse_orders = {}

# =========================
# ADD ITEMS FUNCTION
# =========================

def add_items(item_name, item_cost, quantity_shipped):

    warehouse_orders[item_name] = {
        'item_cost': item_cost,
        'quantity_shipped': quantity_shipped
    }

# =========================
# CALCULATE WAREHOUSE TOTAL
# =========================

def calculate_warehouse_order_total():

    warehouse_orders_total = 0

    for item_name in warehouse_orders:

        item_cost = warehouse_orders[item_name]['item_cost']

        quantity_shipped = warehouse_orders[item_name]['quantity_shipped']

        warehouse_orders_total += (
            item_cost * quantity_shipped
        )

    return warehouse_orders_total

# =========================
# APPLY DISCOUNT
# =========================

def apply_discount(warehouse_orders_total):

    if warehouse_orders_total > 1000:

        return warehouse_orders_total * 0.90

    return warehouse_orders_total

# =========================
# CALCULATE DELIVERY CHARGE
# =========================

def calculate_delivery_charge(
    discounted_warehouse_orders_total
):

    return discounted_warehouse_orders_total * 0.05

# =========================
# CALCULATE FINAL TOTAL
# =========================

def calculate_final_warehouseorders_total(
    discounted_warehouse_orders_total,
    delivery_charge
):

    return round(
        discounted_warehouse_orders_total +
        delivery_charge,
        2
    )

# =========================
# SAVE RECEIPT IMAGE
# =========================

def save_receipt(receipt_lines):

    transaction_description = input(
        "\nEnter transaction description: "
    )

    filename = f"{transaction_description}.png"

    width = 700

    line_height = 35

    padding = 20

    height = (
        len(receipt_lines) * line_height
    ) + (padding * 2)

    # Create image

    image = Image.new(
        "RGB",
        (width, height),
        "white"
    )

    draw = ImageDraw.Draw(image)

    # Load font safely

    try:

        font = ImageFont.truetype(
            "arial.ttf",
            20
        )

    except:

        font = ImageFont.load_default()

    # Starting position

    y_position = padding

    # Draw text line by line

    for line in receipt_lines:

        draw.text(
            (20, y_position),
            line,
            fill="black",
            font=font
        )

        y_position += line_height

    # Save image

    image.save(filename)

    print(f"\nReceipt saved as {filename}")

# =========================
# PRINT RECEIPT
# =========================

def print_receipt():

    print("\n---- RECEIPT ----")

    receipt_lines = []

    receipt_lines.append("==== WAREHOUSE RECEIPT ====")

    # Financial calculations

    warehouse_orders_total = (
        calculate_warehouse_order_total()
    )

    discounted_warehouse_orders_total = (
        apply_discount(warehouse_orders_total)
    )

    delivery_charge = calculate_delivery_charge(
        discounted_warehouse_orders_total
    )

    final_total = (
        calculate_final_warehouseorders_total(
            discounted_warehouse_orders_total,
            delivery_charge
        )
    )

    # Print items

    for item_name, order_details in warehouse_orders.items():

        item_cost = order_details['item_cost']

        quantity_shipped = (
            order_details['quantity_shipped']
        )

        item_total = (
            item_cost * quantity_shipped
        )

        receipt_line = (
            f"{item_name.upper()} | "
            f"£{item_cost} x {quantity_shipped} "
            f"= £{round(item_total, 2)}"
        )

        print(receipt_line)

        receipt_lines.append(receipt_line)

    # Summary section

    print("\n----------------------")

    summary_lines = [

        f"Subtotal: £{round(warehouse_orders_total, 2)}",

        f"After Discount: £{round(discounted_warehouse_orders_total, 2)}",

        f"Delivery Charge (5%): £{round(delivery_charge, 2)}",

        f"Final Total: £{round(final_total, 2)}"
    ]

    # Print and save summary lines

    for line in summary_lines:

        print(line)

        receipt_lines.append(line)

    print("======================")

    # Save receipt image

    save_receipt(receipt_lines)

# =========================
# USER INPUT LOOP
# =========================

while True:

    item_name = input("\nEnter Item Name: ")

    if item_name.lower() == "done":

        break

    item_cost = float(
        input("Enter cost of Item: ")
    )

    quantity_shipped = int(
        input("Enter quantity shipped: ")
    )

    add_items(
        item_name,
        item_cost,
        quantity_shipped
    )

# =========================
# RUN PROGRAM
# =========================

print_receipt()
```
